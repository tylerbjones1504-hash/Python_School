```python
# Prompt user for customer type
customer_type = input("Enter customer type: ")
if customer_type == "VIP":
    discount = 0.2
elif customer_type == "Premium":
    discount = 0.15
else:
    discount = 0.05
print("Customer Type:", customer_type, ", Discount:", discount)
cart = float(input("Enter cart value: "))
final_price = cart - (cart * discount)
print(f"Final price after applying discount: ${final_price:.2f}")
```

```python
prices = [29.99, 49.99, 19.99, 89.99]

discounted = [price * (1 - discount) for price in prices]
print("Discounted Prices:", discounted)
```
